```python
def reverse_word_function(sentence: str) -> str:
    """
    Reverse the order of words in a sentence.

    Args:
        sentence (str): Input sentence.

    Returns:
        str: Sentence with words reversed.

    Raises:
        TypeError: If input is not a string.
    """
    if not isinstance(sentence,str):
        raise TypeError("Input must be string")
    words_list = []
    word = ""
    for char in sentence:
        if char != " ":
            word = word + char
        else:
            words_list.append(word)
            word = ""
    words_list.append(word)

    reverse_sen = ""
    for i in range(len(words_list)-1,-1,-1):
        reverse_sen = reverse_sen + words_list[i]
        if i != 0:
            reverse_sen = reverse_sen + " "
    return reverse_sen
if __name__ == "__main__":
    try:
        user_input = str(input("Enter the word: "))
        result = reverse_word_function(user_input)
        print(result)

    except TypeError as e:
        print("Error: ", e)
    except Exception as e:
        print("Error:", e)
```
