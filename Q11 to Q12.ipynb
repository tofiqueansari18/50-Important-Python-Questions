```python
#Q11. Write a recursive function to calculate the factorial of a given number.

def factorial(n):
    if n == 0 or n == 1:
        return 1

    return n * factorial(n-1)
```

```python
factorial(6)
```

```python
def my_factorial(number : int)-> int:
    """
    Calculate the factorial of a number using recursion.  

    Parameters:
        number (int): A non-negative integer

    Returns:
        int: Factorial of number

    Raises:
        ValueError: If number is negative
        TypeError: If number is not an integer
    """
    if not isinstance(number,int):
        raise TypeError("input must be integer")
    if number < 0:
        raise ValueError("input number must be positive")
    
    if number == 0 or number == 1:
        return 1

    return number * factorial(number-1)
if __name__ == "__main__":
    
    try:
        user_input = int(input("enter the number"))
        number = user_input
        result = my_factorial(number)
        print(f"Factorial of {number} is {result}")
    except (TypeError, ValueError) as error:
        print("Error", error)
    except Exception as error:
        print("something went wrong", error)
        
```

```python
def count_character(input_string: str)-> dict:
    """
    Function to count the occurrences of each character in a string.

    Args:
        input_string (str): The string whose characters are to be counted.

    Returns:
        dict: A dictionary with characters as keys and their counts as values.

    Raises:
        TypeError: If input is not a string.
    """
    if not isinstance(input_string,str):
        raise TypeError("input must be a string")

    char_count = {}
    for char in input_string.lower().replace(" ",""):
        if char in char_count:
            char_count[char] += 1
        else:
            char_count[char] = 1
    return char_count

if __name__ == "__main__":
    try:
        user_input = str(input("enter the words: "))
        input_string = user_input
        result = count_character(input_string)
        print(result)
    except TypeError as error:
        print(error)
    except Exception as error:
        print(error)
```
