FIND COMMON ELEMENTS IN THREE LISTS

Write a function to Find Common Elements in Three Lists

```python
import ast

def find_common_elements(list1: list, list2: list, list3: list) -> list:
    """Returns a list of elements comman in all three lists with out sets"""
    if not isinstance(list1,list) or not isinstance(list2,list) or not isinstance(list3,list):
        raise TypeError("All input values should be a list")
    common = []
    for item in list1:
        if item in list2 and item in list3 and item not in common:
            common.append(item)
    return common

if __name__ == "__main__":
    try:
        user_list1 = ast.literal_eval(input("Enter your first list"))
        user_list2 = ast.literal_eval(input("Enter your second list"))
        user_list3 = ast.literal_eval(input("Enter your third list"))
        user_output = find_common_elements(user_list1,user_list2,user_list3)
        print(user_output)

    except TypeError as e:
        print("Error: ", e)
    except Exception as e:
        print("Something went wrong", e)
```
