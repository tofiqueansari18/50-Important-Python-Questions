```python
def rev_string(input_string: str)-> str:
    """
    Fumction to reverse a given string using for loop.
    Args:
        input_str(str): The string to be reversed.
    Returns:
        str: The reversed version of input_string.
    Raises:
        TypeError: If input is not a string.
    
    """
    if not isinstance(input_string,str):
        raise TypeError("Input must be an alphabetical character")
    reverse_string = ""
    for char in input_string:
        reverse_string = char + reverse_string
    return reverse_string
    
if __name__ == "__main__":
    try:
        user_input = input("Enter your word to reverse: ")
        result = rev_string(user_input)
        print(f"Your reversed word is here:{result}")
    except TypeError as te:
        print(te)
    except Exception as e:
        print(e)
    
    
```

```python
def larg_num(input_list: list[float])-> float:
    
    """
    This function find the largest number from the list
    Args:
        input_list:input_list: list[float]) A list containin numeric values (int or float)
    Returns:
        float: It returns the largest number from the list.
    Raises:
        TypeError: If input is not a list or contain non-numeric values.
        Valueerror: If the list is Empty
    """
    if not isinstance(input_list,list):
        raise TypeError("Input must be a list of numbers float or int")
    if not input_list:
        raise ValueError("List cannot be empty")

    for num in input_list:
        if not isinstance(num,(int, float)):
            raise TypeError("All element must be numeric")

    largest = input_list[0]
    for num in input_list:
        if num > largest:
            largest = num
    return largest

if __name__ == "__main__":
    try:
        user_list = input("Enter list seperat by space in form of number and float only")
        numbers_list = [int(item) for item in user_list.split()]
        result = larg_num(numbers_list)
        print("Largest number in the list is :", result)
    except (TypeError,ValueError) as error:
        print(error)
    except Exception as e:
        print(e)
        
    
```

```python
def is_prime(number:int)->bool:
    """
    Function to check whether a number is prime or not.

    Args:
        number(int): The number to check
        
    Returns:
        Bool: True the number is prime, False otherwise.

    Raiese:
        TypeError: If input is not an integer.
        ValueError: If number is less than 2 (since prime are >=2).
    
    """
    #validate input an integer
    if not isinstance(number,int):
        raise TypeError("Input must be integer")
    #number should be atlest 2
    if number <2:
        raise ValueError("Prime number are greater than or equals to 2")

    for i in range(2, int(number**.5) +1):
        if number % i == 0:
            return False
            break

    else:
        return True

            

#main execution
if __name__ == "__main__":
    try:
        user_input = input("enter the number to checks its prime or not")
        #validate and convert input to integer
        if not user_input.strip().isdigit():
            raise ValueError("input must be a digit")
        number = int(user_input)
        #calling function
        if is_prime(number):
            print(f"{number} you gave is prime number")
        else:
            print(f"{number} you gave is not prime number")

    except (TypeError, ValueError) as error:
        print(error)
    except Exception as e:
        print("something went wrong",e)
        
```

```python
num = 5
factorial = 1
for i in range(1,num+1):
    factorial = factorial * i
print(factorial)
```

```python
def factorial_cal(num: int)-> int:
    """
    Calculate the factorial of a non-negative integer:

    Parameters:
        num(int): The number of which factorial is required.
    Returns:
        int: Factorial of the given number.
    Raises:
        valueError: If number is negative.
        TypeError : If number is not an integer.
    
    
    """
    if isinstance(num,bool):
        raise TypeError("Boolian values are not accepted")

    if not isinstance(num,int):
        raise TypeError("Input must be integer")
    if num<0:
        raise ValueError("Factorial is not defined or zero")

    factorial = 1
    for i in range(1,num+1):
        factorial = factorial * i
    return factorial

if __name__ == "__main__":
    try:
        user_input = input("enter the number to find factorial")
        number = int(user_input)
        print("Factorial:",factorial_cal(number))
    except ValueError as error:
        print("input must be a positive number")
    except TypeError as error:
        print(error)
    except Exception as erroe:
        print("Error",error)
        
```
