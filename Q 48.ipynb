CHECK IF STRING CONTAINS ONLY DIGITS WITHOUT USING BUILTIN METHODS

Write a function that checks if a string contains only numeric characters, without using built-in methods (like str.isdigit()).

```python
def is_all_digits(text: str) -> bool:
    """ Parameters
    ----------
    text : str
        The text to check.

    Returns
    -------
    bool
        True  -> every character is a digit
        False -> at least one character is not a digit
        
    """
    if not isinstance(text, str):
        raise TypeError("Input must be a string.")
    if text == "":
        raise ValueError("Input string cannot be empty.")
    
    for char in text:
        if char < "0" or char > "9":
            return False
    return True

if __name__ == "__main__":
    try:
        test_string = input("Enter a string to check if it contains only digits: ")
        result = is_all_digits(test_string)
        if result:
            print("The string contains only digits.")
        else:
            print("The string does not contain only digits.")
    except TypeError as e:
        print("Error",e)
    except ValueError as e:
        print("Error",e)
    except Exception as e:
        print("An unexpected error occurred:", e)
                
        


       
```
