Generate Lists Of Even Number

```python
n = 20
even_number_list = []
for i in range(2,n+1,2):
    even_number_list.append(i)
print(even_number_list)
```

```python
even_number_list = [i for i in range(2,n+1,2)]
print(even_number_list)
```

```python
def even_list_generator(n: int) -> int:
    """Function to generate a list of even numbers from 1 to n.

    Args:
        n (int): Upper limit

    Returns:
        list: List of even numbers from 1 to n

    Raises:
        TypeError: If input is not an integer
        ValueError: If n is less than 1
    """
    if not isinstance(n, int):
        raise TypeError("Input must be Integer")
    if n<=2:
        raise ValueError("Input must be greater than and equals to 2")

    even_number_list = [i for i in range(2,n+1,2)]
    return even_number_list

#___main__Execution_____

if __name__ == "__main__":
    try:
        user_input = int(input("Enter the number to generate even number list: "))
        result = even_list_generator(user_input)
        print("Even number list is here: ",result)

    except (TypeError,ValueError) as e:
        print("Error :",e)
    except Exception as e:
        print("Error :", e)
    

    
```
