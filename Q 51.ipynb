```python
def total_bal(accounts: dict) -> tuple:
    """
    Return (total_balance, customer_count) for customers whose age is
    strictly greater than `min_age`.

    Raises:
        TypeError:  accounts is not a dict, an account is not a dict,
                    or age/balance/min_age is not a number.
        KeyError:   an account is missing the "age" or "balance" key.
        ValueError: age is negative, or age/balance is NaN or infinite.
    """

    if not isinstance(accounts, dict):
        raise TypeError("accounts must be a dict")
    if len(accounts) == 0:
        raise ValueError("accounts must not be empty")
    total = 0
    count = 0
    for details in accounts.values():
        if details["age"] > 40:
            total += details["balance"]
            count += 1
    return total, count 
if __name__ == "__main__":
    try:
        accounts = {
                "acc1": {"name": "Riya",  "age": 45, "balance": 85000},
                "acc2": {"name": "Karan", "age": 29, "balance": 40000},
                "acc3": {"name": "Neha",  "age": 50, "balance": 120000}}
        result = total_bal(accounts)
        print(f"Tatal amount above 40 is {result[0]}")
        print(f"Total number of customers above 40 is {result[1]}")

    except (TypeError, KeyError, ValueError) as e:
        print(f"Error: {e}")
    except Exception as e:
        print(f"Unexpected error: {e}")


    


```
