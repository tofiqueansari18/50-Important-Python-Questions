FIND THE MEDIAN OF A LIST NUMBER

Write a function to calculate the median of a list numbers.

```python
num_list = [9,5,4,3,2,6,7,1,8]
num_list.sort()
print(num_list)
```

```python
list_lenght = len(num_list)
print(list_lenght)
mid = list_lenght // 2
print(mid)

if list_lenght%2 != 0:
    print(num_list[mid])
else:
    print((num_list[mid - 1] + num_list[mid])/2)
    
```

```python
import ast

def find_median(nums_list: list) -> int:
    if not isinstance(nums_list,list):
        raise ValueError("Input must be a list")
    if len(nums_list) <= 2:
        raise ValueError("Please provide atleast thee values to find median")
    for item in nums_list:
        if not isinstance(item,(int,float)):
            raise ValueError("Items should be integer or float")

    nums_list.sort()
    list_lenght = len(nums_list)
    mid = list_lenght // 2
    if list_lenght % 2 != 0:
        return(nums_list[mid])
    else:
        return(nums_list[mid-1] + nums_list[mid])/2

if __name__ == "__main__":
    try:
        user_list = ast.literal_eval(input("Enter your list to find the median"))
        result = find_median(user_list)
        print(result)

    except (TypeError,ValueError) as e:
        print("Error: ",e)
    except Exception as e:
        print("Something went wrong")
```
