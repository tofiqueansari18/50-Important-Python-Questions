FIND LACM OF TWO NUMBERS

Write a function  to find LCM of two numbers

```python
num1 = 9
num2 = 12
def lcm_finder(num1, num2):
    greatest = max(num1,num2)
    i = 1
    while True:
        lcm = greatest * i
        if lcm %  num1 == 0 and lcm % num2 == 0:
            print("LCM",lcm)
            break
        i = i + 1
lcm_finder(num1,num2)
```

```python
num1 = 123947394739847983749387493894
num2 = 1893739842797298437982749827349827

def lcm_find(num1: int,num2: int) -> int:
    """This function provide LCM of two numbers"""
    if not isinstance(num1,int) or not isinstance(num2,int):
        raise TypeError("Num1 and Num2 should be Integers")
    if num1 == 0 or num2 == 0:
        raise ValueError("LCM not defined for zeros")
    def gcd(num1,num2):
        while num2 != 0:
            num1,num2 = num2, num1 % num2
        return num1
    return abs(num1*num2) // gcd(num1, num2)

if __name__ == "__main__":
    try:
        user_num1 = int(input("Enter your number one"))
        user_num2 = int(input("Enter your number two"))
        result = lcm_find(user_num1,user_num2)
        print(result)

    except (TypeError,ValueError) as e:
        print("Error:", e)
    except Exception as e:
        print("Something went wrong")

lcm_find(num1,num2)
```
