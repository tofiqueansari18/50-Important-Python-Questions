SWAP TWO VARIABLES WITHOUT TEMPORARY VARIABLE

Write a program to swap the values of two variables without using a temporary variable.

```python
def swap_string(string1: str, string2: str) -> str:
    string1, string2 = string2, string1
    return string1, string2

if __name__ == "__main__":
    try:
        user_string1= str(input("Enter the first string"))
        user_string2 = str(input("Enter the secind string"))
        print("String 1 before swap: ", user_string1)
        print("String 2 before swap: ", user_string2)
        user_string1 , user_string2 = swap_string(user_string1,user_string2)
        print("String 1 after swap: ",user_string1)
        print("String 2 after swap: ",user_string2)
        

    except Exception as e:
        print("Something went wrong")
```
