FIND THE LONGEST WORD IN A SENTENCE

Write a function that finds and returns the longest word in a given sentence.

```python
text = """Building a modern requires patience. Is programming difficult? Many people find the process rewarding. Success comes to those who practice daily."""
```

```python
import string
def longest_word_finder(text: str) -> list:
    """ Function that finds longest word in a sentence"""
    if not isinstance(text,str):
        raise TypeError("Input must be string")
    string.punctuation
    text = text.lower()
    pattern = str.maketrans("","",string.punctuation)
    clean_text = text.translate(pattern)
    words_list = clean_text.split()
    word_count_dic = {}
    for word in words_list:
        if word not in word_count_dic:
            word_count_dic[word] = len(word)
    max_len =max(word_count_dic.values())

    longest_word = [key for key,value in word_count_dic.items() if value == max_len]
    return longest_word
if __name__ == "__main__":
    try:
        user_sentence = str(input("Enter your sentences"))
        user_result = longest_word_finder(user_sentence)
        print(user_result)

    except TypeError as e:
        print("Error:",e)
    except Exception as e:
        print("something went wrong:", e)
    


```
