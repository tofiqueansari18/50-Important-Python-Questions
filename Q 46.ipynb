REMOVE NTH OCCURRENCE OF A CHARACTER FROM A STRING

Write a function that removes the nth occurrence of a character from a string.

```python
text = "apples are good"
```

```python
def remove_nth_occurence(string: str, char_to_remove: str, n:int) -> str:
    if not isinstance(string,str) or not isinstance(char_to_remove,str) or not isinstance(n,int):
        raise TypeError("Input and character to remove should be string and n should be integer")
    occurence_count = 0
    result_characters = []
    for current_character in string:
        if current_character == char_to_remove:
            occurence_count = occurence_count + 1
            if occurence_count == n:
                continue
        result_characters.append(current_character)
    return "".join(result_characters)

if __name__ == "__main__":
    try:
        user_string = str(input("Enter the words to remove the alphabet"))
        user_char = str(input("Enter the character to remove"))
        user_nth = int(input("Enter the number of nth character to remove from words"))
        print(remove_nth_occurence(user_string,user_char,user_nth))

    except TypeError as e:
        print("Errror",e)
    except Exception as e:
        print("Something went wrong")
```
