FIND PAIRS WITH GIVEN SUM IN A LIST

Write a function to find all pairs in list that sum u to a given number.

```python
List = [1,2,3,4,5,6]
target = 7
pairs = []
for i in range(len(List)):
    for j in range(i + 1,len(List)):
        if List[i] + List[j] == target:
            pairs.append((List[i],List[j]))
print(pairs)
```

```python
import ast

def find_pairs(List: list, target: int) -> list:
    if not isinstance(List,list):
        raise TypeError("Its should be a list not other")
    if not isinstance(target,(int,float)):
        raise TypeError("Target should be a integer or float")
    if target <= 0:
        raise ValueError("Target should be postitive rather 0 and negative")
    for x in List:
        if not isinstance(x,(int,float)):
            raise TypeError(f"{x} is invalid all elements must be int or float")
    
    pairs = []
    for i in range(len(List)):
        for j in range(i + 1,len(List)):
            if List[i] + List[j] == target:
                pairs.append((List[i],List[j]))
    return pairs

if __name__ == "__main__":
    try:
        user_list = ast.literal_eval(input("Enter your list here"))
        user_target = int(input("Enter your target here"))
        result = find_pairs(user_list,user_target)
        print(result)

    except (TypeError,ValueError) as te:
        print("Error:",te)
    except Exception as te:
        print("Something went wrong",te)

```
