CONVERT LIST OF TOUPLES TO DICTIONARY

Write a function to convert a list of tuples into dictinary

```python
data = [('a', 1), ('b', 2), ('c', 3),('d',4)]
```

```python
def list_to_dic(List: list) -> dict:
    if not isinstance(List,list):
        raise TypeError("Value should be in list")
    my_dict = {}
    for key,value in List:
        my_dict[key] = value
    return my_dict
if __name__ == "__main__":
    data = [('a', 1), ('b', 2), ('c', 3),('d',4)]
    try:
        print(list_to_dic(data))
    except TypeError as e:
        print(e)
    except Exception as e:
        print(e)


```
