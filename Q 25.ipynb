CALCULATE SUM OF A LIST OF NUMBERS USING RECUSRSION

Write a function to calcualte the sum of list of numbers

```python
import ast


def recursive_sum(list1: list) -> int | float:
    """
    Calculates the sum of numeric values in a list using recursion.

    Args:
        list1 (list): A list containing numeric values.

    Returns:
        int | float: The sum of all values in the list.

    Raises:
        TypeError: If the input is not a list or contains non-numeric values.
        ValueError: If the list is empty.
    """
    if not isinstance(list1, list):
        raise TypeError("Input should be a list")

    if len(list1) == 0:
        raise ValueError("List cannot be empty")

    if not isinstance(list1[0], (int, float)):
        raise TypeError("List can contain only numeric values")

    if len(list1) == 1:
        return list1[0]

    return list1[0] + recursive_sum(list1[1:])


if __name__ == "__main__":
    try:
        user_list = ast.literal_eval(input("Enter the list: "))
        user_result = recursive_sum(user_list)
        print(user_result)

    except (TypeError, ValueError, SyntaxError) as e:
        print("Error:", e)

    except Exception as e:
        print("Something went wrong:", e)
```
