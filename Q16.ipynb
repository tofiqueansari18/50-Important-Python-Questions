```python
def calculate_power(base_int: int, power_int:int) -> int:
    """
    Calculate power of a number without using ** operator.[cite: 1]

    Args:[cite: 1]
        base_value (int): Base number[cite: 1]
        power_value (int): Power (must be non-negative)[cite: 1]

    Returns:[cite: 1]
        int: base_value raised to the power power_value[cite: 1]

    Raises:[cite: 1]
        ValueError: If power_value is negative[cite: 1]
        TypeError: If base_value or power_value is not an integer[cite: 1]
    """
    if not isinstance(base_int,int) or not isinstance(power_int,int):
        raise TypeError("Both Base Integer and Power Integer should be Integer")
    if power_int < 0:
        raise ValueError("Power Integer should be greater than 0")

    result = 1
    for x in range(power_int):
        result = result * base_int
    return result
# main Execution
if __name__ == "__main__":
    try:
        base_int = int(input("Provide your base integer: "))
        power_int = int(input("Provide your power integer: "))
        result1 = calculate_power(base_int, power_int)
        print(result1)
    except (TypeError,ValueError) as e:
        print("There is an Error:", e) 
    except Exception as e:
        print("Error: ", e)
```
