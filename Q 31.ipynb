```python
def sum_of_n_number(n: int) -> int:
    """
    Return the sum of squares from 1 to n (1² + 2² + ... + n²).

    Uses the formula: n(n + 1)(2n + 1) // 6

    Raises:
        TypeError: if n is not an integer.
        ValueError: if n < 1.
    """
    if not isinstance(n,int):
        raise TypeError("Input should be an Integer")
    if n < 1:
        raise ValueError("Integer should be a positive number")
    
    total = 0
    for i in range(1,n+1):
        return (n*(n+1)*(2*n+1)//6)

    if __name__ == "__main__":
        try:
            user_n = int(input("Enter the number to get sum of squares"))
            result1 = sum_of_n_number(user_n)
            print(result1)
        
        except (TypeError,ValueError) as e:
            print("Error:",e)
        except Exception as e:
            print("Something went wrong: ",e)
```
