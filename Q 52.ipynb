FIND THE FIST DUPLICATE ELEMENT IN A LIST.

```python
import ast

def first_duplicate(numbers: list) -> int:
    """
    Find the first duplicate element in a list.

    "First duplicate" means the value whose SECOND appearance comes
    earliest while reading the list from left to right.

    Example:
        [10, 5, 3, 4, 3, 5, 6]  ->  3
        (3 repeats at index 4, 5 repeats at index 5, so 3 wins)
    """
    if not isinstance(numbers, list):
        raise TypeError("Input must be a list.")
    if len(numbers) < 2:
        raise ValueError("List must contain at least two elements.")
    
    seen = set()
    for number in numbers:
        if number in seen:
            return number
        seen.add(number)
    
if __name__ == "__main__":
    try:
        user_list = ast.literal_eval(input("Enter a list of numbers (e.g., [10, 5, 3, 4, 3, 5, 6]): "))
        result = first_duplicate(user_list)
        if result is not None:
            print(f"The first duplicate element is: {result}")
        else:
            print("No duplicates found.")
    
    except (TypeError, ValueError) as e:
        print(f"Invalid input: {e}")
    except Exception as e:
        print(f"An error occurred: {e}")
```
