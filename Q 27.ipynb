MERGE TWO SORTED LISTS

Write a function to merge two sorted lists into a single sorted list.

```python
list1 = [4,34,56,78,90]
list2 = [3,56,89,90,1000,4000,5000,6000]
merged_list = []
i = j = 0
while i < len(list1) and j < len(list2):
    if list1[i] <= list2[j]:
        merged_list.append(list1[i])
        i = i + 1
    else:
        merged_list.append(list2[j])
        j = j + 1

#add remaining elements from list1 and list2
while i < len(list1):
    merged_list.append(list1[i])
    i = i + 1
while j < len(list2):
    merged_list.append(list2[j])
    j = j + 1
print(merged_list) 
```

```python
import ast

def merge_sorted_list(list1: list, list2: list) -> list:
    if not isinstance(list1, list) or not isinstance(list2, list):
        raise TypeError("Input should be a list")
    for item in list1 + list2:
        if not isinstance(item,(int,float)):
            raise TypeError("Lists must contain only numbers")

    merged_list = []
    i = j = 0
    while i < len(list1) and j < len(list2):
        if list1[i] <= list2[j]:
            merged_list.append(list1[i])
            i = i + 1
        else:
            merged_list.append(list2[j])
            j = j + 1

    #add remaining elements from list1 and list2
    while i < len(list1):
        merged_list.append(list1[i])
        i = i + 1
    while j < len(list2):
        merged_list.append(list2[j])
        j = j + 1
    return merged_list

if __name__ == "__main__":
    try:
        user_list1 = ast.literal_eval(input("Enter the first sorted list"))
        user_list2 = ast.literal_eval(input("Enter the second sorted list"))
        result1 = merge_sorted_list(user_list1,user_list2)
        print(result1)
    
    except TypeError as e:
        print("Error: ", e)
    except Exception as e:
        print("Something went wrong", e)
        
```
