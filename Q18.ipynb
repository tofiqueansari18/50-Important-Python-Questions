Binary Search Implementation

Write a function that performs binary search on a sorted list and returns the index of the target element or -1 if not found.

```python
sorted_list = [2,5,8,12,16,23,38,45,56,67,78,89,90,95,100]
target = 5
low = 0
high = len(sorted_list) -1
while low <=high:
    mid = (low + high)//2
    if sorted_list[mid] == target:
        print(mid)
        break
    elif target < sorted_list[mid]:
        high = mid -1
    else:
        low = mid + 1
else:
    print(-1)
```

```python
def binary_search(sorted_list: list, target:int ) -> int:
    """
    Function to perform binary search on a sorted list.

    Args:
        sorted_list (list): A sorted list of integers.
        target (int): The element to search for.

    Returns:
        int: Index of the target element if found, otherwise -1.

    Raises:
        TypeError: If input is not a list or target is not an integer.
    """
    if not isinstance(sorted_list,list):
        raise TypeError("Sorted list should be a list")
    if not isinstance(target, int):
        raise TypeError("Target should be an Integer")

    low = 0
    high = len(sorted_list) -1
    while low <=high:
        mid = (low + high)//2
        if sorted_list[mid] == target:
            return mid 
        elif target < sorted_list[mid]:
            high = mid -1
        else:
            low = mid + 1
    else:
        return -1   

if __name__ == "__main__":
    try:
        user_list = [
            int(value)
            for value in input("Enter numbers separated by spaces: ").split()
        ]
        user_list.sort()

        user_target = int(input("Enter the number you are searching for: "))
        result = binary_search(user_list, user_target)
        print("Index:", result)

    except TypeError as e:
        print("eroor: ", e)
    except Exception as e:
        print("error: ", e)


```
