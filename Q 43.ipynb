CONVERT BINARY INTO DECIMAL

Write a function to convert a binary number (as  a string) to decimal.

```python
binary = "1011"
int(binary,2)
```

```python
def binary_to_decimal(binary_str):
    total = 0
    power = 0
    for digit in reversed(binary_str):
        if digit == "1":
            total = total + 2**power
        power = power + 1
    return total

binary_to_decimal("1011")
```

```python
binary_to_decimal("1011101")
```
