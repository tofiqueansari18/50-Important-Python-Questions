CALCULATE GCD GREATEST COMMOM DIVISOR

Write a function to find Greatest Common Divisor(GCD) of two numbers

```python
a = 48
b = 6
smallest_number = min(a,b)
print(smallest_number)
for i in range(smallest_number,0,-1):
    if a % i == 0 and b % i == 0:
        print("gcd = ", i)
        break
```

```python
def gcd_calculator(a: int, b: int) -> int:
    """
    Function to calculate GCD of two numbers.

    Args:
        a (int): First number
        b (int): Second number

    Returns:
        int: Greatest Common Divisor of a and b

    Raises:
        TypeError: If inputs are not integers
        ValueError: If both numbers are zero
    """
    if not isinstance(a,int) or not isinstance(b, int):
        raise TypeError("Input must be Integer")
    if a == 0 and b == 0:
        raise ValueError("GCD is not defined for both numbers because of Zero")
    a,b = abs(a), abs(b)

    while b != 0:
        a,b = b,a%b
    return a
    
#main execution
if __name__ == "__main__":
    try:
        user_a = int(input("Enter your first number"))
        user_b = int(input("Enter your second number"))
        result = gcd_calculator(user_a,user_b)
        print("GCD:",result)

    except TypeError as e:
        print("Error: ", e)
    except ValueError as e:
        print("Error:", e)
    except Exception as e:
        print("Something went wrong:", e)
    
```
