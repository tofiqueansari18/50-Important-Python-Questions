SORT DICTIONARY BY VALUE

Write a function to sort a dictionary by its value.

```python
data = {'apple': 5, 'banana': 2, 'cherry': 8, 'date': 3}
data.items()
```

```python
sorted(data.items())
```

```python
example = [('apple', 5), ('banana', 2), ('cherry', 8), ('date', 3)]
sorted(example,key=lambda x:x[1])
```

```python
import ast

def sort_dict(data: dict) -> dict:
    if not isinstance(data, dict):
        raise TypeError("Input should be in dictionary")
    return sorted(data.items(),key= lambda x:x[1], reverse=False)

if __name__ == "__main__":
    try:
        user_data = ast.literal_eval(input("Enter the dictionary to sort by value"))
        result = sort_dict(user_data)
        print(result)

    except TypeError as e:
        print("Error: ",e)
    except Exception as e:
        print("Something went wrong")

```
