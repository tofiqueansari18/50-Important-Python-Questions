CHECK IF LIST SORTED

Write a function that checks if a list is sorted in ascending order

```python
list1 = [12, 13, 45, 67, 89, 90]
for i in range(len(list1)-1):
    if list1[i] > list1[i+1]:
        print(False)
        break
else:
    print(True)
```

```python
import ast

def is_sorted(list1: list) -> bool:
    """ Function to check if a list is sorted or not """
    if not isinstance(list1, list):
        raise TypeError("Input should be a list")

    for i in range(len(list1)-1):
        if not isinstance(list1[i], int) or not isinstance(list1[i + 1], int):
            raise TypeError("All elements must be Integer")
        if list1[i] > list1[i+1]:
            return False
    return True

if __name__ == "__main__":
    try:
        user_list = ast.literal_eval(input("Enter the sorted list to check"))
        result = is_sorted(user_list)
        print(result)
    
    except TypeError as e:
        print("Errot:",e)
    except Exception as e:
        print("Error:",e)

```
