```python
def is_anagram(string1: str, string2: str) -> bool:
    """
    Check whether two strings are anagrams using character counting.
    Spaces are ignored and comparison is case-insensitive.

    Parameters:
        string1 (str): First input string
        string2 (str): Second input string

    Returns:
        bool: True if strings are anagrams, False otherwise

    Raises:
        TypeError: If either input is not a string
    """
    if not isinstance(string1, str) or not isinstance(string2, str):
        raise TypeError("input must be string")

    string1 = string1.replace(" ", "").lower()
    string2 = string2.replace(" ", "").lower()

    if len(string1) != len(string2):
        return False

    char_count = {}
    for char in string1:
        if char in char_count:
            char_count[char] += 1
        else:
            char_count[char] = 1

    for char in string2:
        if char not in char_count or char_count[char] == 0:
            return False
        char_count[char] -= 1  # 1. Count kam karna zaroori hai

    return True  # 2. Loop ke bahar True return karna zaroori hai


if __name__ == "__main__":
    string1 = "race"
    string2 = "care"
    try:
        result = is_anagram(string1, string2)
        if result:
            print("Your string is anagram")
        else:
            print("Your string is not anagram")
    except TypeError as e:
        print(e)
    except Exception as e:
        print(e)
```
