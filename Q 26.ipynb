SORT A LIST (BUBBLE SORT METHOD)

Write a Function To sort a list using Bubble sort Method

```python
my_list = [96,68,93,55,86,42,37,92,87,50]
lenght = len(my_list)
for i in range(lenght):
    for j in range(0,lenght-i-1):
        if my_list[j] > my_list[j+1]:
            my_list[j], my_list[j + 1] = my_list[j + 1], my_list[j]
print(my_list)
```

```python
import ast


def sort_list(my_list: list) -> list:
    """
    Sort a list of integers in ascending order using the Bubble Sort algorithm.

    Parameters:
    my_list (list): A list containing only integers.

    Returns:
    list: The sorted list in ascending order.

    Raises:
    TypeError:
        - If the input is not a list.
        - If any element in the list is not an integer.
    """
    if not isinstance(my_list, list):
        raise TypeError("Input should be a list of integers")

    for item in my_list:
        if not isinstance(item, int) or isinstance(item, bool):
            raise TypeError("All elements of the list should be integers")

    length = len(my_list)

    for i in range(length):
        for j in range(0, length - i - 1):
            if my_list[j] > my_list[j + 1]:
                my_list[j], my_list[j + 1] = my_list[j + 1], my_list[j]

    return my_list


if __name__ == "__main__":
    try:
        my_list = ast.literal_eval(input("Enter the list to sort: "))
        result1 = sort_list(my_list)
        print(result1)

    except (TypeError, ValueError, SyntaxError) as e:
        print("Error:", e)

    except Exception as e:
        print("Something went wrong:", e)
```
