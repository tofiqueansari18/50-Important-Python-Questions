Queue Concept
A Queue follows the FIFO principle (First In First Out)

```python
enqueue : queue.enqueue(10)
enqueue : queue.enqueue(20)
enqueue : queue.enqueue(30)
Peek : queue.peek()     #Expected Output : 10
dequeue: queue.dequeue()  #Expected Output : 10
dequeue: queue.dequeue()  #Expected Output : 20
dequeue: queue.dequeue()  #Expected Output : 30

```

```python
class Queue:
    def __init__(self):
        self.queue = []
    def is_empty(self):
        return len(self.queue) == 0
    def enqueue(self,value):
        self.queue.append(value)
        print(f"{value} enqueued. Queue {self.queue}")
    def dequeue(self):
        if self.is_empty():
            print("Queue is empty, can not be dequeue")
        else:
            removed = self.queue.pop(0)
            print(f"{removed} dequeue. Queue {self.queue}")
    def peek(self):
        if self.is_empty():
            print("Queue is empty")
        else:
            print(f"Front element : {self.queue[0]}")
```

```python
q = Queue()
q.is_empty()

q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.peek()

q.dequeue()
q.peek()

q.dequeue()
q.dequeue()
q.is_empty()
```
