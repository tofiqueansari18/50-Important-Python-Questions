FIND THE FIRST NON-REPEATED CHARACTER IN A STRING

Write a function to return the first non-repeated character in a string

```python
text = 'swwissn'
char_count = {}
for char in text:
    if char in char_count:
        char_count[char] += 1
    else:
        char_count[char] = 1
print(char_count)

for char in text:
    if char_count[char] == 1:
         print(f"The first non-repeating character is: {char}")
         break
    
```

```python
def non_repeating_character(text):
    char_count = {}
    for char in text:
        if char in char_count:
            char_count[char] += 1
        else:
            char_count[char] = 1

    for char in text:
        if char_count[char] == 1:
            return char
    return None
```

```python
non_repeating_character(text)
```
