REVERSE INTEGER

Write a function that takes an integer and returns its digit reversed.

```python
num = 1234
print(num%10)
print(num//10)
```

```python
def reverse_integer(num: int) -> int:
    if not isinstance(num,int):
        raise TypeError("Value should be an Integer")
    
    sign = -1 if num < 0 else 1
    num = abs(num)
    reversed_num = 0
    while num > 0:
        digit = num%10
        reversed_num = reversed_num * 10 + digit
        num = num//10
    return sign * reversed_num

    #main execution

if __name__ == "__main__":
    try:
        user_integer = int(input("Enter the value of reverse"))
        result1 = reverse_integer(user_integer)
        print(result1)

    except TypeError as e:
        print("Error: ", e)
    except Exception as e:
        print("Something went wrong:", e)
```
