```python
def is_leap_year(year: int) -> bool:
    """
    Check whether a given year is a leap year.

    Conditions:
    1. Divisible by 4 and not divisible by 100
    2. OR divisible by 400

    Args:
        year (int): Year to check

    Returns:
        bool: True if leap year, False otherwise
    raise TypeError:
        if year is not integer
    raise ValueError:
        if year is less than 1582
    """
    if not isinstance(year, int):
        raise TypeError("Year must be an integer value")
    if year < 1582:
        raise ValueError("Year must be greater than 1582")

    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        return True
    else:
        return False

#____main Execution______

if __name__ == "__main__":
    try:
        year = int(input("Enter the year: "))
        result = is_leap_year(year)
        print(f"{year} is {'a leap year' if result else 'not a leap year'}")
    except (ValueError,TypeError) as error:
        print("Error:", error)
    except Exception as error:
        print("Unexpected Error:", error)
```
