```python
#Q6. Palindrome Check (String Problem)
```

```python
string = "madam"
rev_str = string[::-1]
print(string == rev_str)
```

```python
def ch_palindrome(text: str)-> bool:
    """
    Check weather a given string is a palindrome.

    Returns:
        Return True if string is palindrome False otherwise

    Raises:
        TypeError: If input is not a string.
        Value Eroor: If input strin is empty
    """
    if not isinstance(text,str):
        raise TypeError("Input must be string")
    if text.strip() == "":
        raise ValueError("Input string cannot be empty")
    else:
        text.strip().lower()
    reve_str = ""
    for char in text:
        reve_str = char + reve_str
    return text == reve_str

if __name__ == "__main__":
    try:
        user_input = input("enter the palindrome to check: ").strip().lower()
        if ch_palindrome(user_input):
            print(f"{user_input} is a palindrome")
        else:
            print(f"{user_input} is not a palindrome")
    except (TypeError,ValueError) as e:
        print("Error",e)
    except Exception as error:
        print(error)
        
```

```python
def count_vowe(text:str)-> int:
    """
    Count the number of vowels in the given string.

    Parameters:
        text(str): The input string.
    Returns:
        Int: Number of the vowels in the string
    Raises:
        If input is not a string
    """
    if not isinstance(text,str):
        raise TypeError("Input must be a string")
    vowels = "aeiou"
    count = 0
    for char in text.lower():
        if char in vowels:
            count = count + 1
    return count

if __name__ == "__main__":
    try:
        user_input = input("enter the word to count")
        result = count_vowe(user_input)
        print(f"Total vowels: {result}")
    except TypeError as e:
        print("Error",e)
    except Exception as e:
        print(e)
```

```python
def sum_digits(n:int)->int:
    """
    Calculate the sum of digits of a non-negative integer.

    Parameters:
        n (int): The number whose digits will be summed.

    Returns:
        int: Sum of the digits.

    Raises:
        TypeError: If input is not an integer.
        ValueError: If input is negative.
    
    """
    if not isinstance(n,int):
        raise TypeError("input must be integer")
    if n < 0:
        raise ValueError("input must be positive")

    total = 0
    while n > 0:
        digit = (n%10)
        total = total + digit
        n = n // 10
    return (total)

if __name__ == "__main__":
    num = input("enter the number: ")
    try:
        num = int(num)
    except ValueError:
        print("Error","Please enter valid integer only")
    else:
        try:
            result = sum_digits(num)
            print(f"sum of digits is {result}")
        except (TypeError, ValueError) as error:
            print("Error", error)
        except Exception as error:
            print(error)
```
