```python
#Q14 Write a function that prints the first n numbers in the Fibonacci sequence.
# Use an iterative approach.
```

```python
n = 20
a = 0
b = 1
for i in range(n):
    print(a, end = " ")
    a, b = b, a+b
```

```python
def fbnc_iterative(n: int) -> None:
    """
    Print Fibonacci sequence using iterative approach.

    Fibonacci Sequence:
        Each number is the sum of the previous two numbers.
        Example: 0 1 1 2 3 5 8 ...

    Args:
        n (int): Number of terms to print

    Raises:
        ValueError: If n is less than or equal to 0
    """
    if not isinstance(n,int):
        raise TypeError("n must be an integer")
    if n < 0:
        raise ValueError("n must be less than 0")

    a = 0
    b = 1
    for i in range(n):
        print(a, end = " ")
        a, b = b, a+b

if __name__ == "__main__":
    n = 12
    try:
        fbnc_iterative(n)
    except (TypeError,ValueError) as e:
        print("Error: ", e)
    except Exception as e:
        print("Error", e)
```

```python
fbnc_iterative(6)
```
