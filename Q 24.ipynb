REMOVE ALL WHITESPACES FROM A STRING

Write a function that removes all whitespaces characters(spaces, tabs, newlines) from a given string also extra white space.

```python
text = "   \n         This   \t      is \r          python      "
result = ""
previous_space = False
for char in text.strip():
    if not char.isspace():
        result = result + char
        previous_space = False
    else:
        if previous_space == False:
            result = result + " "
            previous_space = True
print(result)
```

```python
def normal_whitespace(text:str) -> str:
    """
    Normalizes whitespace in a string by:
    - Removing leading and trailing whitespace
    - Replacing multiple spaces, tabs, or newlines with a single space

    Args:
        text (str): Input string containing extra whitespace

    Returns:
        str: String with normalized whitespace

    Raises:
        TypeError: If the input is not a string
    """
    if not isinstance(text,str):
        raise TypeError("Input should be a string")

    result = ""
    previous_space = False
    for char in text.strip():
        if not char.isspace():
            result = result + char
            previous_space = False
        else:
            if previous_space == False:
                result = result + " "
                previous_space = True
    return result

#main execution
if __name__ == "__main__":
    try:
        user_text = str(input("Enter the text to remove whitespace"))
        removed = normal_whitespace(user_text)
        print(removed)

    except TypeError as e:
     print("Error: ",e)
    except Exception as e:
        print("Something went wrong: ", e)
```
