```python
def flatten_list(nested_list: list) -> list:
    result = []
    temp_list = list(nested_list)
    while temp_list:
        item = temp_list.pop()
        if isinstance(item,list):
            temp_list.extend(item)
        else:
            result.append(item)
    return result[::-1]
if __name__ == "__main__":
    try:
        print(flatten_list([1,[2,[3,4],5],6]))

    except Exception as e:
        print("Error: ",e)


```
