CHECK IF A NUMBER IS A POWER OF 2

Write a function that checks whether a given number is a power of 2 and returns the exponent (the number of times 2 is multiplied by itself)

```python
n = 16
def power_of_two(n: int) -> bool:
    if n <= 0:
        return False,None
    power = 0
    while n % 2 == 0:
        n = n // 2
        power = power + 1
    if n == 1:
        return True,power
    else:
        return False

if __name__ == "__main__":
    try:
        user_input = int(input("Enter your number to check"))
        result = power_of_two(user_input)
        print(result)
    except Exception as e:
        print("Error:",e)
```
