```python
import string
def frequency_count():
    with open(r"C:\Users\Tofique ansari\OneDrive\Desktop\Practise by Own\Rohit Sir Youtube\Q29 practise.txt","r") as file:
        text = file.read()
    pattern = str.maketrans("","",string.punctuation)
    clean_text = text.translate(pattern)
    words_list = clean_text.lower().split()

    words_count_list = {}
    for word in words_list:
        if word not in words_count_list:
            words_count_list[word] = 1
        else:
            words_count_list[word] = words_count_list[word] + 1

    for key,vlaue in words_count_list.items():
        print(key,vlaue)
```
